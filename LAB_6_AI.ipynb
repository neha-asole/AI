```python
# Name : Neha Asole
# USN No : CS23011
# Practical : Alpha Beta Pruning

import math


def alphabeta(depth, nodeIndex, maximizingPlayer, values, alpha, beta):

    if depth == 3:
        return values[nodeIndex]

    if maximizingPlayer:
        best = -math.inf


        for i in range(2):

            val = alphabeta(depth + 1, nodeIndex * 2 + i,
                            False, values, alpha, beta)

            best = max(best, val)
            alpha = max(alpha, best)

            # Beta cut-off
            if beta <= alpha:
                break

        return best

    # Minimizing player's turn
    else:
        best = math.inf

        for i in range(2):

            val = alphabeta(depth + 1, nodeIndex * 2 + i,
                            True, values, alpha, beta)

            best = min(best, val)
            beta = min(beta, best)


            if beta <= alpha:
                break

        return best

if __name__ == "__main__":

    print("Alpha Beta Pruning Implementation\n")

    values = [3, 5, 6, 9, 1, 2, 0, -1]


    result = alphabeta(0, 0, True, values, -math.inf, math.inf)

    print("The optimal value is :", result)
```
